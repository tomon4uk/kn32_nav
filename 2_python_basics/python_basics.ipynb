## 1. Типи даних і змінні

Створюємо значення різних типів та виводимо значення й тип кожної змінної. Для множини перетворюємо число на рядок, щоб коректно об'єднати його з `a`.

```python
a = "змінна з текстом"
a1 = 'a'
b = 1
b1 = 1.1
c = ["a", 1, 1.25, "Слово", a]
d = {"a": "Слово", "b": 1, a: b}
e = ("a", a)
f = {"ss", a + str(b)}

for variable_name, value in [("a", a), ("a1", a1), ("b", b), ("b1", b1), ("c", c), ("d", d), ("e", e), ("f", f)]:
    print(f"{variable_name} = {value!r}; тип: {type(value).__name__}")
```

## 2. Вбудовані константи та ключові слова

`True`, `None` та `Ellipsis` є вбудованими константами. Модуль `keyword` дає змогу отримати зарезервовані слова Python програмно.

```python
import keyword

print("Константи:", True, None, Ellipsis)
print(f"True має тип {type(True).__name__}, None має тип {type(None).__name__}")
print("Ключові слова Python:")
print(", ".join(keyword.kwlist))
```

## 3. Вбудовані функції

Перевіряємо `abs()` для модуля числа, `len()` для довжини колекції та `repr()` для текстового представлення рядка.

```python
print(abs(-12.5), f"дорівнює {abs(12.5)}; результати рівні: {abs(-12.5) == abs(12.5)}")
print(f"Довжина слова 'Python': {len('Python')}")
print(f"repr рядка: {repr('рядок з\nновим рядком')}")
```

## 4. Цикли

Цикл `for` проходить за елементами списку, а `while` повторюється, доки лічильник менший за 3. Блок `else` циклу `for` виконується після звичайного завершення циклу.

```python
letters = ["a", "b", "c"]
for position, letter in enumerate(letters):
    print(f"На позиції {position} знаходиться буква {letter}")
else:
    print("Цикл for завершився без break")

counter = 1
while counter <= 3:
    print(f"Лічильник: {counter}")
    counter += 1
```

## 5. Розгалуження

Умови `if` / `elif` / `else` обирають гілку відповідно до значення змінної. Додатковий приклад визначає, чи число парне.

```python
temperature = 18
if temperature >= 25:
    print("Тепло")
elif temperature >= 15:
    print("Комфортна температура")
else:
    print("Прохолодно")

number = 7
print(f"{number} парне" if number % 2 == 0 else f"{number} непарне")
```

## 6. Обробка помилок

Ділення на нуль викликає `ZeroDivisionError`. `except` перехоплює конкретну помилку, а `finally` виконується незалежно від того, чи виникла помилка.

```python
denominator = 0
try:
    print(10 / denominator)
except ZeroDivisionError as error:
    print(f"Перехоплено помилку: {error}")
finally:
    print("Блок finally виконався")
```

## 7. Контекстний менеджер `with`

`with` відкриває файл і автоматично закриває його після виходу з блоку. Для прикладу записуємо рядки в тимчасовий текстовий файл, читаємо їх і перевіряємо, що файл закрито.

```python
from pathlib import Path

sample_path = Path("with_example.txt")
with sample_path.open("w", encoding="utf-8") as file:
    file.write("Перший рядок\nДругий рядок")

with sample_path.open("r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"{line_number})> {line.rstrip()}")
print(f"Файл закрито після виходу з with: {file.closed}")

sample_path.unlink()
```

## 8. Lambda-функції

Lambda-функція — це коротка функція-вираз без окремого імені через `def`; вона може приймати аргументи та повертає результат одного виразу. Тут функція множить число на себе, а вбудований `sorted()` використовує lambda як ключ сортування.

```python
square = lambda value: value ** 2
print(f"Квадрат числа 6: {square(6)}")

names = ["Оля", "Андрій", "Іван"]
print("Імена за довжиною:", sorted(names, key=lambda name: len(name)))
```

AI запропонував обчислити суму парних чисел від 1 до 10. Генератор списку відбирає числа, що діляться на 2 без остачі, а `sum()` додає їх.

```python
even_numbers = [number for number in range(1, 11) if number % 2 == 0]
print("Парні числа:", even_numbers)
print("Сума парних чисел:", sum(even_numbers))
```
